# DBLP (co-authorship, local files)

Counts only: number of nodes and hyperedges from your pickled data.

```python
from pathlib import Path
import pickle

dblp_dir = Path("..") / "data" / "trad_data" / "coauthorship" / "dblp"
if not dblp_dir.exists():
    raise FileNotFoundError(dblp_dir.resolve())

with open(dblp_dir / "features.pickle", "rb") as f:
    features = pickle.load(f)
with open(dblp_dir / "hypergraph.pickle", "rb") as f:
    hypergraph_raw = pickle.load(f)

n_nodes = features.shape[0]
n_edges = len(hypergraph_raw)
print(f"n_nodes = {n_nodes}")
print(f"n_hyperedges = {n_edges}")
```
