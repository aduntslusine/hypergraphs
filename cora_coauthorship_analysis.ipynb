# Cora Co-authorship Analysis

Load the co-authorship Cora dataset (not cocitation) and inspect core files.

```python
from pathlib import Path
import pickle
import numpy as np

# Co-authorship path
cora_dir = Path('/Users/lusineadunts/Documents/omms/Hilary/dissertation/data/trad_data/coauthorship/cora')
if not cora_dir.exists():
    raise FileNotFoundError(f'Co-authorship Cora folder not found: {cora_dir}')

with open(cora_dir / 'features.pickle', 'rb') as f:
    features = pickle.load(f)
with open(cora_dir / 'labels.pickle', 'rb') as f:
    labels = np.asarray(pickle.load(f))
with open(cora_dir / 'hypergraph.pickle', 'rb') as f:
    hypergraph_raw = pickle.load(f)

splits_dir = cora_dir / 'splits'
split_files = sorted([p for p in splits_dir.glob('*.pickle')])
splits = []
for p in split_files:
    with open(p, 'rb') as f:
        splits.append(pickle.load(f))

print(f'Loaded Cora co-authorship: n_nodes={features.shape[0]}, n_features={features.shape[1]}, n_edges_raw={len(hypergraph_raw)}')
print(f'Loaded {len(splits)} split(s)')
```
